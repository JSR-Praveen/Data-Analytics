```python
file_path= "Supermart Grocery Sales - Retail Analytics Dataset.csv"
df = pd.read_csv(file_path)
print("Basic Info:")
print(df.info())
print("\nSummeary Statictics:")
print(df.describe())
```

```python
print(df.head())
df['Total'] = df[['Sales','Discount','Profit']].sum(axis=1)
```

```python
city_sales = df.groupby('City')['Sales'].sum().sort_values(ascending = False)
print(city_sales)
```

```python
plt.Figure(figsize=(15,10))
city_sales.plot(kind='bar',color='red')
plt.title('Total sales by city')
plt.xlabel('city')
plt.ylabel('Total Sales')
plt.xticks(rotation= 45)
plt.show()
```

```python
import random
df['Gender'] = random.choices(['Male','Female'], weights=[0.5,0.5],k= len(df))
```

```python
gender_count = df['Gender'].value_counts()
print(gender_count)
```

```python
plt.Figure(figsize=(30,30))
gender_count.plot(kind='pie',autopct='%1.1f%%',colors=['pink','lightblue'])
plt.title('Gender Distributio')
plt.ylabel('')
plt.show()
```
